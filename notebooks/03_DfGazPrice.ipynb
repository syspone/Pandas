## 3. Caminho do arquivo e colunas de data

O notebook roda a partir da **própria pasta** (`notebooks/`), que é o padrão do Antigravity,
do VS Code e do JupyterLab. Por isso usamos `..` para subir um nível até a raiz do projeto

As colunas de data são informadas em `parse_dates` para já entrarem como

```python
ARQUIVO = Path("../Dataset/GazPrice.csv")
COLUNAS_DATA = ["DATA INICIAL", "DATA FINAL"]
```

## 4. Carregando o dataset

`sep=","` e `encoding="utf-8"` já são o padrão do `read_csv`, então não precisam ser passados.

### Atribuindo Constantes

```python
produto_view = df_gaz['PRODUTO']                                    # No pandas 3 (Copy-on-Write), funciona como uma cópia independente.
produto_view
```

```python
produto_copy_bkp = df_gaz['PRODUTO'].copy()                        # Cópia explícita: no pandas 3 dá o mesmo resultado da célula acima.
produto_copy_bkp    
```

```python
df_gaz['PRODUTO'] = 'Combustivel'                                   # Substitui o dado para "combustivel"
df_gaz.head()
```

### Por que `produto_view` não mudou? (Copy-on-Write)

Repare que, mesmo depois de `df_gaz['PRODUTO'] = 'Combustivel'`, **as duas variáveis
continuaram com `ETANOL HIDRATADO`**.

Isso acontece por causa do **Copy-on-Write** ("copia ao escrever"), que é o padrão desde o **pandas 3.0**:

- Selecionar uma coluna (`df_gaz['PRODUTO']`) **funciona como uma cópia**. Por baixo dos panos,
  o pandas só copia os dados de verdade no momento em que alguém tenta alterar, o que economiza memória.
- Por isso, **alterar o DataFrame nunca muda uma variável separada antes**, e alterar a variável
  nunca muda o DataFrame.
- Para **separar** uma coluna, o `.copy()` deixou de ser necessário. Ele continua útil para deixar
  a intenção clara no código ou se você usar uma versão antiga do pandas (antes da 3.0), em que
  `produto_view` podia, sim, ser afetada.

> Regra prática no pandas 3: para mudar o DataFrame, altere **o próprio DataFrame**
> (`df_gaz['PRODUTO'] = ...` ou `df_gaz.loc[linha, 'PRODUTO'] = ...`), e não uma variável que veio dele.
